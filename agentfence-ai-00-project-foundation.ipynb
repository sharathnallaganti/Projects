# AgentFence.AI — Module 0: Project Foundation

## Project Overview

AgentFence.AI is a consumer-focused security gateway based on the idea of a **Personal Intent Firewall**. It compares what a person wants to do with the access a website, application, or artificial intelligence (AI) assistant requests and the actions it later performs.

The central question is:

> Does this access or action fit what the user asked to do?

## Problem

People may approve access without understanding its full scope. For example, someone wants to upload one résumé, but a website asks to read and modify their entire cloud drive, access their contacts, or keep ongoing access to their account.

AgentFence.AI aims to explain that mismatch before the person approves it.

## Intended Users and Benefits

The first users are individual consumers completing tasks in supported browser workflows. AgentFence.AI aims to help them:

- Understand what an access request allows.
- Compare the request with the task they want to complete.
- Choose narrower or shorter access when available.
- Notice supported actions that go beyond what they approved.
- Keep a clear record of their decisions and observed activity.

## How the Project Works

1. The user states a purpose, such as “Upload one résumé.”
2. A website, application, or AI assistant requests access or proposes an action.
3. AgentFence.AI compares the request with the user’s purpose.
4. It explains any mismatch in plain language.
5. The user chooses to allow, narrow, or reject the request.
6. For supported workflows, later actions are compared with the approved scope.
7. A security receipt records the purpose, decision, and observed actions.

## Example

**User’s purpose:** Upload one résumé.

**Request:** Read and modify every file in the user’s cloud drive, access contacts, and retain ongoing account access.

**AgentFence.AI response:** Explain that these permissions are broader than the stated task. If the service supports it, offer a narrower choice, such as selecting one résumé for a one-time upload.

## Core Project Functions

- Capture or receive the user’s intended task.
- Represent requested permissions and actions in a consistent format.
- Compare user intent with the requested access and action scope.
- Explain potential consequences in plain language.
- Recommend reduced or time-limited access when possible.
- Flag supported actions that exceed the approved purpose.
- Create a security receipt.

## Project Terms

| Term | Meaning |
|---|---|
| **Intent** | What the person wants to accomplish. |
| **Request** | Access or permission requested by a website, app, or AI assistant. |
| **Action** | An operation the website, app, or assistant proposes or performs. |
| **Scope** | Which data or resources are included, and whether they can be read, changed, or shared. |
| **Decision** | The project’s recommendation: allow, warn, or block. |
| **Observed event** | An action recorded in a supported or simulated workflow. |
| **Security receipt** | A readable record of the purpose, decision, approved scope, and observed events. |

## Prototype Boundaries

The notebook prototype will use synthetic examples and mock services. It will not use real personal files, credentials, or cloud accounts.

The project will make claims only about browser workflows it can observe or simulate. It will not claim to monitor every phone application, every website, or every action on a device. A browser extension and local service can be considered after the intent-to-action model is defined and evaluated.

## Success Measures

The project will measure:

- Excessive access requests correctly warned about or blocked.
- Simulated disclosures prevented.
- False warnings on legitimate requests.
- User understanding of the explanation.
- Time taken to complete a legitimate task.

## Cost and Privacy Approach

The initial notebook work is designed to cost **$0** by using Kaggle, synthetic data, mock services, and no paid AI application programming interfaces or hosted cloud services. All sample names, file contents, and account details will be fictional.
